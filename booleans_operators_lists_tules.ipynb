```python
#update tuples
#once tuple is created you can't change it. they are unchangeable or immutable.
#that's why convert to list to change the list and then convert back to tuple
x=("apple","banana","cherry")
y=list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
```

```python
#add items

#convert into a list  => 1st way
thistuple = ("apple","banana","cherry")
y=list(thistuple)
y.append("orange")
thistuple=tuple(y)
print(thistuple)

#add tuple to a tuple  => 2nd way
y=("orange",)  #a new tuple with valur orange
thistuple += y
print(thistuple)
```

```python
#remove items
#tuples are unchangeable , so you can't remove items
thistuple=("apple","banana","cherry")
y= list(thistuple)
y.remove("apple")
thituple = tuple(y)
print(thistuple)

#del keyword
del thistuple
#print(thistuple) => error , delete tuple completely 
```

```python
#unpack tuples
fruits = ("apple","banana","cherry")  #packing a tuple => assign values to tuple
(green,yellow,red) = fruits #unpack => allowed to extract the values back into avriable

print(green)
print(yellow)
print(red)
```

```python
#using asterisk * =>no of variable less than no of values, add * to the variable name and 
#the values will be assigned to the variable as a list
fruits = ("apple","banana","cherry","strawberry","raspberry")
(green , yellow, *red) = fruits

print(green); print(yellow); print(red)
```
