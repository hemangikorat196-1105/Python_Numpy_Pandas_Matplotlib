```python
#slicing
b="hello, world!"
print(b[2:5])
```

```python
#slice from the start -> leaving out the start index,range will 
# start from first character
b="hello, world!"
print(b[:5])
```

```python
#slice to the end -> leaving out the end index, the range will go to the end
b="hello, world!"
print(b[2:])
```

```python
#negative index -> to start the slice from the end of the string
b="hello, world!"
print(b[-5:-2])
```

```python
#escape characters -> \ backslash
#to insert characters that are illegal in a string , use an escape characters
txt='it\'s alright.'
print(txt)

txt = "this will insert one \\(backslash)."
print(txt)

txt = "Hello\nWorld"
print(txt)

txt = "Hello\rWorld"
print(txt)

txt = "Hello\tWorld"
print(txt)

txt = "Hello\bWorld"
print(txt)

txt = "Hello\fWorld"
print(txt)

txt = "\110\145\154\154\157"
print(txt)

txt = "\x48\x65\x6c\x6c\x6f"
print(txt)
```

```python
#string methods

txt = "python is FUN!"
x = txt.capitalize()
print (x)

x = txt.casefold()
print (x)

x = txt.encode()
print (x)
```

```python
txt = "python is FUN!"
x = txt.isalnum()
print (x)

x = txt.isalpha()
print (x)

x = txt.isascii()
print (x)

x = txt.isdecimal()
print (x)

x = txt.isdigit()
print (x)
```

```python
txt="python is FUN!"
x = txt.isidentifier()
print (x)

x = txt.islower()
print (x)

x = txt.isnumeric()
print (x)

x = txt.isprintable()
print (x)

x = txt.isspace()
print (x)

x = txt.istitle()
print (x)

x = txt.isupper()
print (x)
```
