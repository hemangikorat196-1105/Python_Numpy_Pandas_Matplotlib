```python
set1={"apple",1,"banana",0,"cherry"}
set2={False,"google",1,"apple",2,True}

set3=set1.intersection(set2)  #only contains the items that are present in both set
print(set3)

set4 = set1.difference(set2)  #will return a new set that will contain only the items from the first set that are not present in the other set
print(set4)

set5 = set1 -set2  # - operator instead of the difference()
print(set5)

set1.difference_update(set2)  #method will keep the items from the first set that are not in the other set,but it will change the original set instead of returning a new set
print(set1)

set1.intersection_update(set2)  #will also keep only the duplicates,but it will change the original set instead of returning a new set
print(set1)
```

```python
set1={"apple","banana","cherry"}
set2={"google","microsoft","apple"}

set3=set1.symmetric_difference(set2)
print(set3)

set4=set1 ^ set2
print(set4)

set1={"apple","banana","cherry"}
set2={"google","microsoft","apple"}

set1.symmetric_difference_update(set2)
print(set1)
```

```python
#immutable version of set 
#like sets => contain unique,unordered,unchangeable elements
#unlike sets => elements cannot be added or removed from a frozenset
x = frozenset({"apple","banana","cherry"})
print(x)
print(type(x))
```

```python
fs = frozenset({1,2,3})
cp = fs.copy()  #returns shallow copy
print(fs); print(cp)
```

```python
a=frozenset({1,2,3,4})
b=frozenset({3,4,5})
print(a.difference(b))   #returns a new frozenset with the difference
print(a-b)

print(a.intersection(b))  #returns a new frozenset with the intersection
print(a & b)
```

```python
a = frozenset({1,2})
b = frozenset({3,4})
c = frozenset({2,3})
print(a.isdisjoint(b))  #returns true if theere is no intersection between two frozensets
print(a.isdisjoint(c))

print(a.issubset(b))  #returns true if this frozenset is a subset of another
print(a <= b)
print(a < b)

print(a.issuperset(b))  #returns true if this frozenset is a superset of another
print(a >= b)
print(a > b)

a = frozenset({1,2,3})
b = frozenset({3,4,5})
print(a.symmetric_difference(b))  #returns a new frozenset with the symmetric difference
print(a ^ b)

print(a.union(b)) #returns a new frozenset containing the union
print(a | b)
```
